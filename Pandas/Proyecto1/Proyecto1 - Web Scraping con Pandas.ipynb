```python
#Crear una variable root
root = "https://www.football-data.co.uk/mmz4281/"
```

```python
#Multiples ligas
-https://www.football-data.co.uk/mmz4281/2425/E0.csv
-https://www.football-data.co.uk/mmz4281/2425/E1.csv
-https://www.football-data.co.uk/mmz4281/2425/E2.csv
-https://www.football-data.co.uk/mmz4281/2425/E3.csv
-https://www.football-data.co.uk/mmz4281/2425/EC.csv
```

```python
#Crear lista de ligas

import pandas as pd

leagues = ['E0', 'E1', 'E2', 'E3', 'EC']
frames = []

# hacer loop en la lista leagues, leer multiples csv y adjuntarlos
for league in leagues:
    df= pd.read_csv(root + '2425' + '/' + league + '.csv')
    frames.append(df)
```

```python
#Tamaño de los frames
len(frames)
```

```python
# mostrar 1er, 2do, y 3er elemento
frames[4]
```

```python

#Crear lista de ligas

import pandas as pd

leagues = ['E0', 'E2', 'E3',]
frames = []

#Hacer loop rango de temporadas e insertar una nueva columna

for league in leagues:
    for season in range(15,21):
        df= pd.read_csv(root + str(season) + str(season+1) + '/' + league + '.csv')
        df.insert(1,'season',season)
        frames.append(df)
```

```python
#Total de frames (6 temporadas x 3 ligas
len(frames)
```

```python
#Mostrar 1er y ultimo elemento (columna season fue creada
frames[17]
```

# Proyecto 1 Parte 3

### Organizar Data en un diccionario

```python
# crear diccionario con nombre original de liga como llave
dict_countries = {
              'Spanish La Liga':'SP1', 'Spanish Segunda Division':'SP2',
              'German Bundesliga':'D1', 
              'English Premier League':'E0', 'English League 1':'E2', 'English League 2':'E3',
             }
```

```python
# Obtener un elemento del diccionario
dict_countries['Spanish La Liga']
```

```python
# Loop sobre diccionario
for league in dict_countries:
    print(league)

for league in dict_countries:
    print(dict_countries [league])
```

```python
#Crear lista de ligas
import pandas as pd

dict_historical_data = {}

#leagues = ['E0', 'E2', 'E3',]

#Hacer loop sobre llaves de diccionario

for league in dict_countries:
    frames = []
    for season in range(15,21):
        df= pd.read_csv(root + str(season) + str(season+1) + '/' + dict_countries[league] + '.csv')
        df.insert(1,'season',season)
        frames.append(df)
    df_concat = pd.concat(frames)
    dict_historical_data[league] = df_concat
```
